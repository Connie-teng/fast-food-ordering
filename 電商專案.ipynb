```python
import random
import math
import time

# ==============================
# 全域變數宣告
# ==============================
store_name = '賣噹噹'
all_members = {}
member_points = {}   # 🆕 會員積分
isMember = False
member_name = ''     # 🆕 記錄登入的會員名稱

A = 'A：吉士堡＋薯條＋可樂'
B = 'B：麥香雞＋冰炫風＋紅茶'
C = 'C：麥香魚＋雞翅＋玉米濃湯'
D = 'D：雞塊＋沙拉＋雪碧'
E = 'E：薯條＋冰炫風＋紅茶'
menu = [A, B, C, D, E]
number0 = ['A', 'B', 'C', 'D', 'E']
price = [150, 140, 130, 90, 80]
num_content = dict(zip(number0, menu))
num_price   = dict(zip(number0, price))

product    = ['吉士堡', '麥香雞', '麥香魚', '冰炫風', '雪碧', '紅茶', '玉米湯', '雞塊', '薯條', '雞翅', '沙拉']
type_price = [90,       95,       75,       45,       30,     40,     55,       60,     30,     55,     65   ]

number = [str(n) for n in range(1, len(product) + 1)]
item   = dict(zip(product, type_price))

total  = 0    # 總金額
bought = {}   # 購物車 {'1': 3, 'A': 2}

# 🆕 今日特餐（隨機）
daily_special_item  = random.choice(product)
daily_special_price = item[daily_special_item] - 20
daily_special_index = product.index(daily_special_item) + 1  # 對應號碼


# ==============================
# 工具函式
# ==============================
def split():
    print("=" * 52)

def slow_print(text, delay=0.03):
    """🆕 逐字打印，增加趣味感"""
    for ch in text:
        print(ch, end='', flush=True)
        time.sleep(delay)
    print()

def show_banner():
    """🆕 開場 ASCII Banner"""
    print("""
╔══════════════════════════════════════════════════╗
║                                                  ║
║    🍔  歡 迎 光 臨  【 賣 噹 噹 】  🍔           ║
║         M·Dong·Dong  Fast Food                   ║
║                                                  ║
╚══════════════════════════════════════════════════╝
    """)

def show_daily_special():
    """🆕 顯示今日特餐"""
    print(f"🌟【今日特餐】🌟  {daily_special_index}. {daily_special_item}  "
          f"原價 ${item[daily_special_item]} → 特價 ${daily_special_price}！")
    split()


# ==============================
# 主選單
# ==============================
def start():
    global isMember, member_name, total, bought
    total  = 0
    bought = {}
    isMember    = False
    member_name = ''

    show_banner()
    show_daily_special()

    while True:
        member_tag = f"[👤 {member_name}]" if isMember else "[訪客]"
        print(f"\n{member_tag}  目前購物車金額：${total}")
        print("┌─────────────────────────────────┐")
        print("│  0. 登入／加入會員               │")
        print("│  1. 🍱 指定套餐                  │")
        print("│  2. 🛒 自選商品（單點）           │")
        print("│  3. 📋 顯示購物車                │")
        print("│  4. 💳 結帳                      │")
        print("│  5. 🚪 離開                      │")
        print("└─────────────────────────────────┘")
        w = input("請輸入選項：").strip()

        if   w == "0": Customer_Membership_system()
        elif w == "1": print_meals(); order_1()
        elif w == "2": type_select()
        elif w == "3": show_bought()
        elif w == "4":
            checkout()
            slow_print("\n🙏  感謝光臨賣噹噹，下次見！")
            break
        elif w == "5":
            print('👋  謝謝光臨，歡迎再來！')
            break
        else:
            print("⚠️  輸入錯誤，請重新選擇")
        split()


# ==============================
# 會員系統
# ==============================
def Customer_Membership_system():
    global isMember, member_name
    print("\n🔐 請進行會員認證")
    name = input("帳號：").strip()

    if name in all_members:
        pw = input("密碼：")
        while True:
            if all_members[name] == pw:
                isMember    = True
                member_name = name
                pts = member_points.get(name, 0)
                slow_print(f"✅  歡迎回來，{name}！您目前有 {pts} 點積分。")
                break
            else:
                print("❌  密碼錯誤！")
                pw = input("請再次輸入密碼：")
    else:
        print("查無此帳號，是否要加入會員？(y/n)")
        if input("→ ").strip().lower() == 'y':
            add_member()
        else:
            print("好的，以訪客身份繼續。")

def add_member():
    global isMember, member_name
    account = input("設定帳號：").strip()
    while True:
        pw1 = input("設定密碼：")
        pw2 = input("再次輸入密碼：")
        if pw1 == pw2:
            all_members[account]   = pw1
            member_points[account] = 0
            isMember    = True
            member_name = account
            slow_print(f"🎉  恭喜 {account} 成功加入會員！享有會員專屬九五折！")
            break
        else:
            print("⚠️  兩次密碼不一致，請重試")


# ==============================
# 套餐區
# ==============================
def print_meals():
    split()
    print("📋  指定套餐一覽：")
    for i, (k, v) in enumerate(zip(number0, menu)):
        print(f"  {v}  💰 ${price[i]}")
    split()

def order_1():
    """
    ✅ 修正：支援連續輸入多個套餐（如 A B C 或 AAB），
    輸入 ok 結束點餐。
    """
    global total
    print("💡  提示：可一次輸入多個套餐代碼（如：A B C 或 AAB），輸入 ok 結束")

    while True:
        raw = input("請選擇套餐（ok 結束）：").strip().upper()
        if raw == 'OK':
            return

        # 支援空格分隔「A B C」或連續輸入「AAB」
        if ' ' in raw:
            choices = raw.split()
        else:
            choices = list(raw)  # 'AAB' → ['A','A','B']

        valid   = [c for c in choices if c in num_content]
        invalid = [c for c in choices if c not in num_content]

        if invalid:
            print(f"⚠️  以下代碼無效，已略過：{', '.join(invalid)}")
        if not valid:
            print("沒有有效的套餐代碼，請重新輸入")
            continue

        # 顯示本次選擇
        print("\n📝  本次選擇：")
        batch_total = 0
        batch_summary = {}
        for c in valid:
            batch_summary[c] = batch_summary.get(c, 0) + 1
        for c, qty in batch_summary.items():
            subtotal = num_price[c] * qty
            print(f"  {num_content[c]}  x{qty}  ${subtotal}")
            batch_total += subtotal
        print(f"  小計：${batch_total}")

        confirm = input("確認加入購物車？(v 確認 / x 取消)：").strip().lower()
        if confirm == 'v':
            for c, qty in batch_summary.items():
                bought[c] = bought.get(c, 0) + qty
                total     += num_price[c] * qty
            print(f"✅  已加入！目前購物車總金額：${total}")
        else:
            print("取消，請重新選擇")


# ==============================
# 單點區
# ==============================
def type_select():
    split()
    print("🛒  單點商品：")
    for i, (name, p) in enumerate(item.items(), 1):
        star = "⭐" if name == daily_special_item else "  "
        tag  = f"（今日特價 ${daily_special_price}！）" if name == daily_special_item else f"  ${p}"
        print(f"  {i:>2}. {star} {name}{tag}")
    print()
    cart_loop()

def cart_loop():
    """
    單點使用 while 迴圈。
    支援頓號「、」、逗號「,」或空格一次輸入多項商品代碼。
    例如：1、3、5  或  1,3,5  或  1 3 5
    """
    global total
    print("💡  提示：可用頓號一次輸入多項，例如：1、3、5")
    while True:
        raw = input("輸入商品代碼（頓號分隔多項，e 結束）：").strip()
        if raw.lower() == "e":
            split()
            return

        # 依序嘗試頓號、逗號、空格來分割；都沒有就視為單一輸入
        if '、' in raw:
            codes = [c.strip() for c in raw.split('、')]
        elif ',' in raw:
            codes = [c.strip() for c in raw.split(',')]
        elif ' ' in raw:
            codes = [c.strip() for c in raw.split()]
        else:
            codes = [raw.strip()]

        # 過濾空字串
        codes = [c for c in codes if c]

        valid   = [c for c in codes if c in number]
        invalid = [c for c in codes if c not in number]

        if invalid:
            print(f"⚠️  以下代碼無效，已略過：{'、'.join(invalid)}")
        if not valid:
            print("沒有有效的商品代碼，請重新輸入")
            continue

        # 逐一詢問數量
        print("\n📝  本次選擇：")
        batch = {}   # {code: (qty, name, price)}
        for c in valid:
            idx  = int(c) - 1
            name = product[idx]
            p    = daily_special_price if name == daily_special_item else type_price[idx]
            t = input(f"  → {c}. {name}  ${p}  購買數量：").strip()
            if not t.isdigit() or int(t) <= 0:
                print(f"  ⚠️  {name} 數量無效，略過此項")
                continue
            batch[c] = (int(t), name, p)

        if not batch:
            print("本次沒有有效數量，請重新輸入")
            continue

        # 顯示小計
        batch_total = sum(qty * p for qty, _, p in batch.values())
        for c, (qty, name, p) in batch.items():
            print(f"     {name} x{qty} = ${qty * p}")
        print(f"  小計：${batch_total}")

        confirm = input("確認加入購物車？(v 確認 / x 取消)：").strip().lower()
        if confirm == 'v':
            for c, (qty, name, p) in batch.items():
                bought[c] = bought.get(c, 0) + qty
                total    += p * qty
            print(f"✅  已加入！目前購物車總金額：${total}")
        else:
            print("取消，請重新選擇")


# ==============================
# 購物車
# ==============================
def show_bought():
    global total
    if not bought:
        print("🛒  購物車是空的！")
        return

    split()
    print("🧾  購物車明細：")
    for key, qty in bought.items():
        if key in number:
            idx  = int(key) - 1
            name = product[idx]
            p    = daily_special_price if name == daily_special_item else type_price[idx]
            print(f"  [{key}] {name}  x{qty}  = ${p * qty}")
        elif key in number0:
            print(f"  [{key}] {num_content[key]}  x{qty}  = ${num_price[key] * qty}")
    print(f"  合計：${total}")
    split()

    j = input("刪除項目請輸入代碼，k 繼續：").strip()
    if j == "k":
        return
    elif j in bought:
        if j in number:
            idx  = int(j) - 1
            name = product[idx]
            p    = daily_special_price if name == daily_special_item else type_price[idx]
            total -= p * bought[j]
        elif j in number0:
            total -= num_price[j] * bought[j]
        del bought[j]
        print(f"🗑️  已刪除，更新後金額：${total}")
    else:
        print("⚠️  購物車沒有這個品項！")


# ==============================
# 🆕 幸運轉盤（取代舊 lottery）
# ==============================
def lucky_wheel():
    global total
    total = math.ceil(total)

    if total == 0:
        print("🤔  甚麼都不買就走了嗎？")
        return

    print("\n🎡  消費滿額自動享折扣：")
    if total >= 1000:
        total = math.ceil(total * 0.85)
        slow_print("🎊  消費破千！獲得 85 折優惠！")
    elif total >= 500:
        total = math.ceil(total * 0.90)
        slow_print("🎉  消費破五百！獲得 9 折優惠！")
    elif total >= 400:
        print(f"💡  再加 ${500 - total} 元就有 9 折折扣！")
    else:
        add = input("還需要加購嗎？（1 是 / 2 否）：").strip()
        if add == '1':
            type_select()

    # 🆕 幸運轉盤動畫
    print("\n🎰  轉動幸運轉盤中", end='', flush=True)
    for _ in range(5):
        time.sleep(0.3)
        print(".", end='', flush=True)
    print()

    prizes = [
        ('🍦  冰炫風買一送一券', None),
        ('🥚  一盒蛋塔券',       None),
        ('💰  現折 50 元',       50  ),
        ('🥤  免費可樂一杯',     None),
        ('🥔  薯餅兌換券',       None),
        ('😅  謝謝參與，下次好運', None),
    ]
    name, discount = random.choice(prizes)
    if discount:
        total = max(0, total - discount)
        slow_print(f"🎁  恭喜獲得：{name}！已自動折抵！")
    else:
        slow_print(f"🎁  恭喜獲得：{name}！")


# ==============================
# 結帳
# ==============================
def checkout():
    global total
    lucky_wheel()

    # 會員折扣
    if isMember:
        discounted = math.ceil(total * 0.95)
        slow_print(f"\n👤  會員九五折：${total} → ${discounted}")
        total = discounted

        # 🆕 計算積分（每消費 10 元 = 1 點）
        pts_earned = total // 10
        member_points[member_name] = member_points.get(member_name, 0) + pts_earned
        print(f"⭐  本次獲得 {pts_earned} 積分，累計：{member_points[member_name]} 點")

    # 付款方式
    print("\n💳  請選擇付款方式：")
    print("  1. 信用卡   2. LINE Pay   3. 現金")
    ans = input("→ ").strip()
    if ans == '1':
        slow_print("💳  信用卡感應中……嗶！付款完成！")
    elif ans == '2':
        slow_print("📱  請掃描 LINE Pay 條碼……付款完成！")
    elif ans == '3':
        slow_print("💵  請至櫃台交付現金，謝謝！")
    else:
        slow_print("⚠️  輸入無效，預設現金結帳。")

    split()
    slow_print(f"🧾  本次消費總計：${total} 元")
    slow_print("🙏  感謝您的光臨，祝您用餐愉快！")


# ==============================
# 主程式進入點
# ==============================
start()
```
